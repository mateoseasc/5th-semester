## Algoritmo de Recocido Simulado


```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go

x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)

X,Y = np.meshgrid(x,y)
Z = -5/(1+X**2+Y**2)+np.sin(np.tan(np.exp(-5/(1+X**2+Y**2))))

fig = go.Figure(data=[go.Surface(z=Z,x=X,y=Y)])
fig.show()
```

```python
import random 

#FUNCIÓN OBJETIVO

def fun_obj(x,y):
    f = -5/(1+x**2+y**2)+np.sin(np.tan(np.exp(-5/(1+x*x*2+y**2))))
    return f 

#FUNCIÓN QUE GENERA LAS SOLUCIONES VECINAS
def solucion_vecina(x,y,perturbacion):
    x_new = x + np.random.uniform(-perturbacion,perturbacion)
    y_new = y + np.random.uniform(-perturbacion,perturbacion)
    return x_new, y_new

#=========================RECOCIDO SIMULADO============================0

def recocido_simulado(x_inicial,y_inicial,T_inicial,alfa,iteraciones,perturbacion):
    #Valores Iniciales 
    x_actual = x_inicial
    y_actual = y_inicial
    f_actual = fun_obj(x_actual,y_actual)
    mejor_x = x_actual
    mejor_y = y_actual
    mejor_f = f_actual
    T = T_inicial
    i = 0

    while i < iteraciones and T >  - float("inf"):
        #Generamos las soluciones vecinas
        x_vecina, y_vecina = solucion_vecina(x_actual,y_actual,perturbacion)
        f_vecina = fun_obj(x_vecina,y_vecina)

        #Calcular Delta

        delta = f_vecina - f_actual

        #Criterio de aceptación
        if delta < 0:
            #Se acepta
            x_actual = x_vecina
            y_actual = y_vecina
            f_actual = f_vecina

        else:
            #Se acepta con probabilidad
            prob = np.exp(-delta/T)
            if random.random() < prob:
                x_actual = x_vecina
                y_actual = y_vecina
                f_actual = f_vecina

        #Actualizar mejor solucion 

        if f_actual < mejor_f:
            mejor_x = x_actual
            mejor_y = y_actual
            mejor_f = f_actual

        #Enfriamiento
        T = alfa*T
        i +=1

    return mejor_x, mejor_y, mejor_f

#=================EJECUCIÓN====================
x0 = np.random.uniform(-5,5)
y0 = np.random.uniform(-5,5)
T0 = 10.0
alfa = 0.8 #Va entre .8 y .99  
iteraciones = 1000
perturbacion = .2

mejor_x,mejor_y,mejor_f = recocido_simulado(x0,y0,T0,alfa,iteraciones,perturbacion)

print(f"MEJOR SOLUCIÓN ENCONTRADA EN {iteraciones} iteraciones")
print("x = ", mejor_x)
print("y = ", mejor_y)
print("f(x,y) = ", mejor_f)
```

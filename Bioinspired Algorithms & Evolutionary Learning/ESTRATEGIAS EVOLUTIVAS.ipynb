# ALGORITO DE ESTRATEGIAS EVOLUTIVAS
Se busca determinar el mínimo de la función de dos variables de Rosenbrock

```python
import random
import math 

#---------------FUNCIONES-----------------

#FUNCION FITNESS
def fun_obj(x1,x2):
    return 100*(x1**2-x2**2)**2 + (1-x1)**2

#FUNCION PARA INICIAR UN INDIVIDUO
def iniciar_individuo(Imin,Imax):
    x1 = random.uniform(Imin,Imax)
    x2 = random.uniform(Imin,Imax)
    sigma1 = random.uniform(0.1,1.0)
    sigma2 = random.uniform(0.1,1.0)
    return[x1,x2,sigma1,sigma2]
    

#MUTACIÓN AUTOADAPTATIVA
def mutacion(individuo,Imin,Imax):
    n = 2 #DIMENSIÓN
    tau = 1/math.sqrt(2*math.sqrt(n))
    tau_prima = 1 / math.sqrt(2*n)

    #Ruido global y local
    N0 = random.gauss(0,1)
    individuo[2] *= math.exp(tau_prima * N0 + tau * random.gauss(0,1))
    individuo[3] *= math.exp(tau_prima * N0 + tau * random.gauss(0,1))

    #Mutación en variables de decision
    individuo[0] += individuo[2] * random.gauss(0,1)
    individuo[1] += individuo[3] * random.gauss(0,1)

    #Restringimos al dominio
    if individuo[0] < Imin:
        individuo[0] = Imin
    if individuo[0] > Imax:
        individuo[0] = Imax
    if individuo[1] < Imin:
        individuo[1] = Imin
    if individuo[1] > Imax:
        individuo[1] = Imax

    return individuo

#SELECCIÓN (mu,lambda)-ES

def seleccion_mu_lambda(poblacion,mu):
    pob_ordenada = sorted(poblacion,key=lambda individuo: fun_obj(individuo[0],individuo[1]))
    return pob_ordenada[:mu]
```

```python
#----------------------------ALGORITMO PRINCIPAL----------------------------
Imin = -50
Imax = 50
mu = 50
lambd = 200
generaciones = 1000
#Se crea poblacion inicial
P = []
for i in range(mu):
    P.append(iniciar_individuo(Imin,Imax))

#EVOLUCIÓN
for g in range(generaciones):
    offspring = []
    for i in range(lambd):
        #Se selecciona un padre
        padre = random.choice(P)
        #Copiamos al individuo
        hijo = padre[:]
        #Mutación
        hijo = mutacion(hijo,Imin,Imax)
        offspring.append(hijo)

    Poblacion_unida = P + offspring

    #Seleccionamos los mejores mu 
    P = seleccion_mu_lambda(Poblacion_unida,mu)
    mejor = P[0]

print("ALGORITMO DE ESTRATEGIA EVOLUTIVA (mu,lambda)")
print("\nMejor Solución encontrada: ",mejor)
print("VALOR DE LA FUNCIÓN", fun_obj(mejor[0],mejor[1]))
```

## PRÁCTICA PSO

```python
import numpy as np

#------------------- Funciones ------------------

#definimos la función objetivo
def bukin(posicion):
    x = posicion[0]
    y = posicion[1]
    return 100*np.sqrt(np.abs(y-0.01*x**2))+0.01*np.abs(x+10)

#Función que inicializa las particulas

def inicializar_praticulas(n_particulas, dim, limites):
    #Iniciamos los arreglos para posicion y velocidad
    posiciones = np.zeros((n_particulas, dim))
    velocidades = np.zeros((n_particulas, dim))

    #Se extraen los limites inferior y superior para cada dimension
    l_inf = np.array([limites[i][0] for i in range(dim)])
    l_sup = np.array([limites[i][1] for i in range(dim)])


    # Se crean de forma aleatoria las posiciones y velocidades de las particulas
    for i in range(n_particulas):
        for d in range(dim):
            posiciones[i][d] = np.random.uniform(l_inf[d],l_sup[d])
            #Para las velocidades iniciales se crean con magnitud 10% del rango de busqueda
            #Practica estándar del PSO
            velocidades[i][d] = np.random.uniform(-abs(l_sup[d]-l_inf[d]),abs(l_sup[d]-l_inf[d]))*0.1

    return posiciones, velocidades

#Ahora sigue el algoritmo
#=============================PSO===========================
#Parámetros
dim = 2
n_particulas = 500
max_iter = 5000
w = 0.7
c1 = 1.5
c2 = 1.5
limites = [(-15,5),(-3,3)]

#Iniciamos valores
posiciones, velocidades = inicializar_praticulas(n_particulas, dim, limites)
pbest_posicion = np.copy(posiciones)
pbest_valores = np.zeros(n_particulas)

for i in range(n_particulas):
    pbest_valores[i] = bukin(posiciones[i])

gbest_indice = np.argmin(pbest_valores)
gbest_posicion = np.copy(pbest_posicion[gbest_indice])
gbest_valor = np.copy(pbest_valores[gbest_indice])

#Limite de volocidad (por dimension)
v_max = np.zeros(dim)
for d in range(dim):
    v_max[d] = 0.2*(limites[d][1]- limites[d][0])

#Evolución por iteraciones

for t in range(max_iter):
    for i in range(n_particulas):
        r1 = np.random.random(dim)
        r2 = np.random.random(dim)

        #Actualizamos la velocidad
        for d in range(dim):
            cognitivo = c1*r1[d]*(pbest_posicion[i][d]-posiciones[i][d])
            social = c2*r2[d]*(gbest_posicion[d]-posiciones[i][d])
            velocidades[i][d] = w*velocidades[i][d]+cognitivo+social


        #Se actualiza la posicion
        for d in range(dim):
            posiciones[i][d] = posiciones[i][d]+velocidades[i][d]

        #Se evaluan las nuevas posiciones
        fitness = bukin(posiciones[i])

        #Se actualiza le pbest
        if fitness < pbest_valores[i]:
            pbest_valores[i] = fitness
            pbest_posicion[i] = np.copy(posiciones[i])

        #Se actualiza el gbest
        if fitness < gbest_valor:
            gbest_valor = fitness
            gbest_posicion = np.copy(pbest_posicion[i])

print("Minimo encontrado:")
print("X=", gbest_posicion[0])
print("Y=", gbest_posicion[1])
print("f(x,y)=", gbest_valor)
```

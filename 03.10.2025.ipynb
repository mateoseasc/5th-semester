# EJERCICIO 20



Descargar precios ajustados.

Calcular rendimientos diarios.

Definir Rf anual (CETES 28 días) y convertir a Rf diaria.

Calcular Sharpe diario y anualizar.

```python
import yfinance as yf
import numpy as np
import pandas as pd

# 1) Datos
serie = yf.download("AMX", start="2024-01-01", end="2025-01-01")["Close"]
rend = serie.pct_change().dropna()

# 2) Parámetros
dias_trading = 252
rf_anual = 0.0719  
rf_diaria = rf_anual / dias_trading

# 3) Métricas (como escalares, sin warnings)
rp = rend.mean().item()
sigma = rend.std(axis=0).item()

sharpe_diario = (rp - rf_diaria) / sigma
sharpe_anual = sharpe_diario * np.sqrt(dias_trading)

print(f"Rend. diario promedio: {rp:.4%}")
print(f"Volatilidad diaria   : {sigma:.4%}")
print(f"Sharpe diario        : {sharpe_diario:.2f}")
print(f"Sharpe anualizado    : {sharpe_anual:.2f}")
```

# EJERCICIO 21

Objetivo: comparar qué acción tuvo mejor rendimiento ajustado por riesgo en 2024 utilizando CETES 91 días como Rf.
Pasos:

Descargar dos tickers de la BMV.

Calcular rendimientos diarios y métricas por activo.

Usar la misma Rf (CETES 91d → anual → diaria) para ambos.

Rankear por Sharpe anualizado.



```python
import yfinance as yf
import numpy as np
import pandas as pd

# Descargar datos
AMX = yf.download("AMX", start="2024-01-01", end="2025-01-01")["Close"]
BIMBOA = yf.download("BIMBOA.MX", start="2024-01-01", end="2025-01-01")["Close"]

activos = [('AMX', AMX), ('BIMBO', BIMBOA)]

# Parámetros
tasa_libre_riesgo_anual = 0.0747  # CETES 91 días
dias_trading = 252
tasa_libre_riesgo_diaria = (1 + tasa_libre_riesgo_anual)**(1/dias_trading) - 1

def funcion(activo):
    rend_diarios = activo.pct_change().dropna()
    rend_promedio_diario = float(rend_diarios.mean())

    vol_diaria = float(rend_diarios.std())
    if vol_diaria == 0:
        sharpe_diario = np.nan
        sharpe_anual = np.nan
    else:
        sharpe_diario = (rend_promedio_diario - tasa_libre_riesgo_diaria) / vol_diaria
        sharpe_anual = sharpe_diario * np.sqrt(dias_trading)

    print(f"Rend. diario promedio  : {rend_promedio_diario:.4%}")
    print(f"Volatilidad diaria     : {vol_diaria:.4%}")
    print(f"Sharpe diario          : {sharpe_diario:.2f}")
    print(f"Sharpe anualizado      : {sharpe_anual:.2f}")

# Ejecutar cálculo por activo
for nombre, activo in activos:
    print(f"\nCálculo para el activo: {nombre}")
    funcion(activo)
    print("\n")
```

```python
import yfinance as yf
import numpy as np
import pandas as pd

serie = yf.download("AMX", start="2024-01-01", end="2025-01-01")["Close"]
rend = serie.pct_change().dropna()

dias_trading = 252
rf_anual = 0.0719  
rf_diaria = rf_anual / dias_trading

sortino_diario = float((rend.mean() - rf_diaria) / rend[rend < 0].std())
sortino_anual = sortino_diario * np.sqrt(dias_trading)

print(f"Sortino diario        : {sortino_diario:.2f}")
print(f"Sortino anualizado    : {sortino_anual:.2f}")
```

# EJERCICIO 23
Objetivo: comparar cuál acción tuvo mejor rendimiento ajustado por pérdidas en 2024 con CETES 91d como Rf.

Pasos:

Descargar dos tickers BMV.

Calcular rendimientos y la downside deviation por activo.

Misma 
R
f
Rf​ (CETES 91d → diaria) para ambos.

Ordenar por Sortino anualizado.



```python
import yfinance as yf
import numpy as np
import pandas as pd

# Descargar datos
AMX = yf.download("AMX", start="2024-01-01", end="2025-01-01")["Close"]
BIMBOA = yf.download("BIMBOA.MX", start="2024-01-01", end="2025-01-01")["Close"]

activos = [('AMX', AMX), ('BIMBO', BIMBOA)]

def funcion(activo):
    rend_diarios = activo.pct_change().dropna()

    tasa_libre_riesgo_anual = 0.0747
    dias_trading = 252
    tasa_libre_riesgo_diaria = (1 + tasa_libre_riesgo_anual)**(1/dias_trading) - 1

    rend_promedio_diario = float(rend_diarios.mean())  # <- convertir a escalar

    # Downside deviation
    diff = rend_diarios - tasa_libre_riesgo_diaria
    downside = diff[diff < 0]
    downside_dev = float(np.sqrt(np.mean(downside**2))) if not downside.empty else 0.0

    if downside_dev == 0:
        sortino_diario = np.nan
        sortino_anual = np.nan
    else:
        sortino_diario = (rend_promedio_diario - tasa_libre_riesgo_diaria) / downside_dev
        sortino_anual = sortino_diario * np.sqrt(dias_trading)

    print(f"Rend. diario promedio  : {rend_promedio_diario:.4%}")
    print(f"Downside deviation     : {downside_dev:.4%}")
    print(f"Sortino diario         : {sortino_diario:.2f}")
    print(f"Sortino anualizado     : {sortino_anual:.2f}")

for nombre, activo in activos:
    print(f"\nCálculo para el activo: {nombre}")
    funcion(activo)
    print("\n")

```

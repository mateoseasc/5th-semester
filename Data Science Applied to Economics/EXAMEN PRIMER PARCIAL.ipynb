## PREGUNTA 14
Con la lista [0.05, -0.02, 0.00, 0.07, -0.01], clasifica cada rendimiento como positivo, negativo o neutro.

```python
import numpy as np 
lista = [0.05, -0.02, 0.00, 0.07, -0.01]
rendimientos_simples = []
for i in range(1, len(lista)):
    if lista[i-1] != 0:
        Rendimiento = (lista[i] - lista[i-1]) / lista[i-1]
    else:
        Rendimiento = 0  
    rendimientos_simples.append(Rendimiento)
for i in rendimientos_simples:
    if i > 0:
        print(f"este rendimiento es POSITIVO {i}")
    elif i == 0:
        print(f"este rendimiento es NEUTRO {i}")
    elif i < 0:
        print(f"este rendimiento es NEGATIVO {i}")
```

## PREGUNTA 27
Escribe un programa en Python que grafique los siguientes rendimientos usando matplotlib: [0.02, 0.03, -0.01, 0.015, 0.04].


```python
import matplotlib.pyplot as plt
rendimientos = [0.02, 0.03, -0.01, 0.015, 0.04]
#Crear gráfico de barras
plt.bar(range(1, len(rendimientos)+1), rendimientos, color='black')
plt.title('Rendimientos Diarios')
plt.xlabel('Días')
plt.ylabel('Rendimiento')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # Línea en y=0
plt.grid(True)
```

## PREGUNTA 28
Usa pandas para convertir la lista [120, 125, 130, 128, 132] en una serie y calcula los rendimientos logarítmicos.

```python
#Cálculo de rendimiento logarítmico con pandas
import pandas as pd 
import numpy as np

#Serie de prcios simulados
precios = pd.Series([120, 125, 130, 128, 132])

#Cálculo de rendimiento logarítmico
rend_log = np.log(precios / precios.shift(1))

#Mostrar resultados
print(f'Rendimientos logarítmicos:\n{rend_log.dropna()}')
```

## PREGUNTA 30
Con los rendimientos [0.08, 0.09, 0.07, 0.12] y una tasa libre de riesgo rf = 0.03, calcula el Ratio de Sharpe.

```python

rendimientos = np.array([0.08, 0.09, 0.07, 0.12],dtype=float)

tasa_libre_riesgo_anual = 0.03
tasa_libre_riesgo_mensual = (1 + tasa_libre_riesgo_anual)**(1/12) - 1  # capitalización compuesta

rend_promedio = rendimientos.mean()
volatilidad = rendimientos.std()  

# Sharpe
sharpe_mensual = (rend_promedio - tasa_libre_riesgo_mensual) / volatilidad
sharpe_anual = sharpe_mensual * np.sqrt(12)

print(f'Sharpe Ratio Mensual: {sharpe_mensual}')
print(f'Sharpe Ratio Anual: {sharpe_anual}')
```

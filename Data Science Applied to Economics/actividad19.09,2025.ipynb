```python
#ACTIVIDAD 16
# Crear un diccionario con activos y sus rendimientos
rendimientos_activos = {
    "AAPL":[ 0.15, .13 , .10],
    "TSLA":[0.08, -.3, .50] ,
    "NFLX": [-0.05, .10, .34]
}

for activo, rendimientos_list in rendimientos_activos.items():
    # Calcular el rendimiento promedio de la lista
    rendimiento_promedio = sum(rendimientos_list) / len(rendimientos_list)

    print(f"Análisis para {activo}:")
    print(f"Rendimientos: {rendimientos_list}")
    print(f"Rendimiento Promedio: {rendimiento_promedio:.2%}")
    
    # Aplicar la lógica de recomendación
    if rendimiento_promedio > 0.10:
        print("Recomendación: Alto rendimiento – Recomendado")
    elif 0 < rendimiento_promedio <= 0.10:
        print("Recomendación: Rendimiento moderado")
    else:
        print("Recomendación: Rendimiento negativo – Riesgoso")
    print("-" * 30)

```

```python
#ACTIVIDAD 19
import yfinance as yf
import numpy as np 
import pandas as pd

# Descargar datos mensuales (intervalo='1mo')
BTC = yf.download('BTC-USD', period='3y', interval='1mo')
AAPL = yf.download('AAPL', period='3y', interval='1mo')
TSLA = yf.download('TSLA', period='3y', interval='1mo')

# Calcular rendimientos mensuales y limpiar NaN
BTC['Rendimiento_Mensual'] = BTC['Close'].pct_change()
BTC_clean = BTC['Rendimiento_Mensual'].dropna()

AAPL['Rendimiento_Mensual'] = AAPL['Close'].pct_change()
AAPL_clean = AAPL['Rendimiento_Mensual'].dropna()

TSLA['Rendimiento_Mensual'] = TSLA['Close'].pct_change()
TSLA_clean = TSLA['Rendimiento_Mensual'].dropna()

# Imprimir los promedios aritméticos
BTC_promedio = BTC_clean.mean()
AAPL_promedio = AAPL_clean.mean()
TSLA_promedio = TSLA_clean.mean()

print('Promedios Aritméticos por activo mensual')
print('BTC: ', f"{BTC_promedio * 100:.2f}%")
print('AAPL: ', f"{AAPL_promedio * 100:.2f}%")
print('TSLA: ', f"{TSLA_promedio * 100:.2f}%")

# Convertir el promedio mensual a anualizado (Aritmético)
BTC_anualizado = (1 + BTC_promedio)**12 - 1
AAPL_anualizado = (1 + AAPL_promedio)**12 - 1
TSLA_anualizado = (1 + TSLA_promedio)**12 - 1

# Imprimir los resultados anualizados
print('\nPromedios Aritméticos Anualizados por activo')
print('BTC Anualizado: ', f"{BTC_anualizado * 100:.2f}%")
print('AAPL Anualizado: ', f"{AAPL_anualizado * 100:.2f}%")
print('TSLA Anualizado: ', f"{TSLA_anualizado * 100:.2f}%")
```

```python
import yfinance as yf
import numpy as np 

# Descargar datos mensuales
BTC = yf.download('BTC-USD', interval='1mo',start='2018-01-01', end='2025-01-01')
BTC_rendimientos = BTC['Close'].pct_change().dropna()

# Tasa libre de riesgo (7% anual)
tasa_libre_riesgo_anual = 0.07
tasa_libre_riesgo_mensual = (1 + tasa_libre_riesgo_anual)**(1/12) - 1  # capitalización compuesta

# Estadísticos
rend_promedio = BTC_rendimientos.mean()
volatilidad = BTC_rendimientos.std(ddof=1)  # desviación estándar muestral

# Sharpe
sharpe_mensual = (rend_promedio - tasa_libre_riesgo_mensual) / volatilidad
sharpe_anual = sharpe_mensual * np.sqrt(12)

print(f'Sharpe Ratio Mensual: {sharpe_mensual}')
print(f'Sharpe Ratio Anual: {sharpe_anual}')
```

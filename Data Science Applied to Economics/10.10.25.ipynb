## Ejercicio 25
 Calcular la volatilidad con fórmula manual

```python
rendimientos = [0.02, 0.03, -0.01, 0.015, 0.04]

def Desviacion_estandar_muestral(rendimientos):
    n = len(rendimientos)
    promedio = sum(rendimientos) / n

    cuadrados = []
    for r in rendimientos:
        cuadrado = (r - promedio)**2
        cuadrados.append(cuadrado)
    sumatoria = sum(cuadrados)

    std = (sumatoria / (n-1)) **(1/2)
    return std

Desviacion_estandar_muestral(rendimientos)
```

## EJERCICIO 26
Usa numpy para calcular la desviación estándar de estos rendimientos:


```python
import numpy as np
rendimientos = [0.05, 0.01, -0.02, 0.03, 0.025]
volatilidad = np.std(rendimientos,ddof = 1)
print (f" La volatilidad es: {volatilidad:.4f}")
```

## EJERCICIO 27
Supón que la volatilidad diaria es de 0.017. Calcula la volatilidad anualizada con 252 días hábiles.


```python
vol_diaria = 0.017
vol_anualizada = vol_diaria * ((252) ** (1/2))
print(f"Volatilidad anualizada: {vol_anualizada:.2%}")
```

```python
import numpy as np

precios = np.array[100,101.5,99.8,100.6,102.0,101.2,103.0], dtype = (float)

returns_simple = precios[1:] / precios[:1] - 1
returns_log = np.log(precios[1:] / precios[:-1])

vol_diaria_simple = np.std(returns_simple,ddof = 1)
vol_diaria_log = np.std(returns_log,ddof = 1)

print ('Volatilidad Diaria Simple: ', vol_diaria_simple)
print ('Volatilidad Diaria logarítmica: ', vol_diaria_log)
```

```python
import yfinance as yf
import pandas as pd 
import numpy as np 

data = yf.download('BTC-USD', start='2019-01-01', end='2021-09-12')
data['Retorno'] = data['Close'].pct_change()
Retorno_promedio_diario = data['Retorno'].mean()
volatilidad_diaria = data['Retorno'].std()
print('Rendimiento Promedio Diario: ', Retorno_promedio_diario)
print('Volatilidad Diario: ', volatilidad_diaria)
```

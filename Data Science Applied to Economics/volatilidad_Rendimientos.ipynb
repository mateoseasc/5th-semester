### Volatilidad Rendimientos


```python
#Calculo de rendimiento simple con pandas
import pandas as pd

#Lista de precios simulados
precios = [100, 105, 110, 108, 112]

#Convertir a serie de pandasseaborn
serie_precios = pd.Series(precios)

#Calcular rendimiento simple
rendimientos_simples = (serie_precios / serie_precios.shift(1)) - 1
rendimientos_simples = rendimientos_simples.dropna()  # Eliminar el primer valor NaN
print(f'Rendimientos simples:\n{rendimientos_simples}')
```

```python

#Cálculo de rendimiento logarítmico con pandas
import pandas as pd 
import numpy as np

#Serie de prcios simulados
precios = pd.Series([100, 105, 110, 108, 112])

#Cálculo de rendimiento logarítmico
rend_log = np.log(precios / precios.shift(1))

#Mostrar resultados
print(f'Rendimientos logarítmicos:\n{rend_log.dropna()}')
```

```python
#Programa Completo 

#Primero definimos las funciones para calcular los rendimientos
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def calcular_rendimiento_simple(precios):
    seriepd = pd.Series(precios)
    rendimientos_simples = (seriepd / seriepd.shift(1)) - 1
    rendimientos_simples = rendimientos_simples.dropna()  
    return rendimientos_simples

def calcular_rendimiento_logaritmico(precios):
    seriepd = pd.Series(precios)
    rendimientos_logaritmicos = np.log(seriepd / seriepd.shift(1))
    rendimientos_logaritmicos = rendimientos_logaritmicos.dropna()  
    return rendimientos_logaritmicos
    
def graficar_rendimientos(rendimientos_log,rendimientos_simples, titulo='Rendimientos Diarios'):
    plt.figure(figsize=(12, 6))
    plt.bar(range(1, len(rendimientos_log)+1), rendimientos_log, color='blue', alpha=0.6, label='Rendimiento Logarítmico')
    plt.bar(range(1, len(rendimientos_simples)+1), rendimientos_simples, color='orange', alpha=0.6, label='Rendimiento Simple')
    plt.title(titulo)
    plt.xlabel('Días')
    plt.ylabel('Rendimiento')
    plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  
    plt.legend()
    plt.grid(True)
    plt.show()
  

#Luego usamos las funciones con una lista de precios simulados
precios = [100, 105, 110, 108, 112,105,1000,500,200,50]
rendimientos_simples = calcular_rendimiento_simple(precios)
rendimientos_logaritmicos = calcular_rendimiento_logaritmico(precios)
graficar_rendimientos(rendimientos_logaritmicos, rendimientos_simples)
rendimientos_simples, rendimientos_logaritmicos
```

```python
#Ejemplo: Cálculo de rendimientos o retornos

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Extraer datis de yahoo finance
data = yf.download('BTC-USD', start='2024-01-01', end='2025-05-13')

#Calcular el retorno potencial o simple diario a partir de cierre
#tail -> muestra las ultimas filas de  un data

data['Retorno'] = data['Close'].pct_change()

#Calcular rendimiento logarítmico diario a partir del precio de cierre
data['LogRetorno'] = np.log(data['Close'] / data['Close'].shift(1))

plt.plot(data['Retorno'], label='Rendimiento Simple', color='blue')
plt.plot(data['LogRetorno'], label='Rendimiento Logarítmico', color='orange')
plt.title('Rendimientos simples vs logarítmicos de BTC-USD')
plt.xlabel('Fecha')
plt.ylabel('USD')
plt.grid(True)

plt.figure(figsize=(8, 4))
sns.histplot(data['Retorno'].dropna(), bins=50, kde=True, color='blue', label='Rendimiento Simple')
sns.histplot(data['LogRetorno'].dropna(), bins=50, kde=True, color='orange', label='Rendimiento Logarítmico')
plt.title('Distribución de Rendimientos de BTC-USD')
plt.xlabel('Rendimiento Diario')
plt.grid(True)
plt.show()

#Exporta rendimientos a CSV
data.to_csv('rendimientos_btc.csv')
```
